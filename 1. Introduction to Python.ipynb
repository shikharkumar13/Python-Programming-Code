# Part Two: What Actually Happens When You Run Code

This is the part most beginners never get explained, but understanding it will make you a far more confident programmer. Let's look under the hood.

### How a Computer Actually Thinks

A computer's processor (the CPU) only understands one language: **machine code**. This is pure binary, just 1s and 0s representing electrical on/off signals. Something like `num = 10` means absolutely nothing to the raw hardware.

So there is a gap. On one side is human-friendly Python. On the other is a machine that only speaks binary. Something has to bridge that gap.

### How Python Reads Your Code

**Step 1: Reading and checking (parsing).** Python reads your code top to bottom and checks that the grammar makes sense. If you forgot a colon, it stops here and reports a syntax error before running anything.

**Step 2: Translating to bytecode.** Python converts your code into a compact intermediate form called **bytecode**. This is not machine code yet. (Those mysterious `.pyc` files and the `__pycache__` folder are saved bytecode, kept so Python does not have to re-translate unchanged code every time.)

**Step 3: Executing line by line.** The **Python Virtual Machine (PVM)** takes that bytecode and runs it one instruction at a time. This is what makes Python *interpreted*: the running and the final translation happen together, live.

So the journey looks like this:

    Your Python code  →  Bytecode  →  Python Virtual Machine  →  CPU does the work

You write the first box. Python handles everything after it, automatically.

### What Happens in Memory

When you write `num = 10`, two separate things happen:

1. Python creates an **object** in memory to hold the value `10`. It sits at a specific address, like a house on a street.
2. The name `num` becomes a **label** that points to that address.

Here is the insight that makes everything else click: in Python a variable is not a box that contains a value. It is a label attached to a value living in memory.

That explains the earlier mystery. `val1 = val2 = val3 = 40` did not make three copies of `40`. It made one object and stuck three labels on it, which is why `id()` returned the same address for all three.

```python
val1 = val2 = val3 = 40
val1 is val2 is val3   # one object, three labels pointing at it
```

And when you reassign a name, the old value is not overwritten in place. A brand-new object is created elsewhere and the label is simply *moved* over to point at it.

```python
num = 10
first_id = id(num)

num = 'Python'         # the label moves to a new object
id(num) == first_id    # False: it is not the same object any more
```

### Garbage Collection: Automatic Cleanup

The old `10` above, with no labels left pointing at it, eventually gets cleaned up automatically. That is **garbage collection**. Python keeps track of how many labels point to each object, and when an object becomes unreachable Python frees that memory for reuse.

In lower-level languages like C you have to claim and release memory manually, and forgetting to do so causes serious bugs. Python does this housekeeping for you.

### Interpreted vs Compiled Languages

Both approaches solve the same problem, translating human code into machine code. They just do it at different times.

A **compiled** language (C, C++, Rust) translates your entire program into machine code *before* it runs, producing a standalone executable. Running it is then very fast, because the translating is already done.

> Compiling is like translating an entire book from French to English, printing the finished version, and handing it to the reader.

An **interpreted** language (Python) translates and runs your code line by line, as it goes, every time you run it. There is no separate build step.

> Interpreting is like having a live translator next to you, translating each sentence out loud the moment it is spoken. More flexible, but a bit slower.

| | Compiled (C, C++) | Interpreted (Python) |
|---|---|---|
| When translation happens | Once, before running | Continuously, while running |
| Speed of execution | Very fast | Slower |
| Ease of use / testing | Slower to test (rebuild each time) | Instant, just run it |
| Catches errors | Many caught before running | Often only when that line runs |
| Portability | Tied to the system it was built for | Runs anywhere Python is installed |

**The honest middle ground:** Python is technically a blend. It first compiles to bytecode, then interprets that bytecode. But from a beginner's point of view the experience is interpreted: write code, run it instantly, no build step.

### Static vs Dynamically Typed Languages

The *type* of a variable means what kind of data it holds. The difference between static and dynamic typing is about **when**, and whether, you have to declare it.

A **statically typed** language (Java, C++) requires you to declare the type up front, and it is locked in:

    // Java
    int age = 25;      // 'age' is an integer, forever
    age = "hello";     // ERROR: you cannot put text in an integer variable

A **dynamically typed** language (Python) works out the type automatically from whatever you assign, and lets you change it freely:

```python
age = 25
print(type(age))

age = 'hello'      # perfectly fine in Python
print(type(age))
```

| | Statically Typed (Java, C++) | Dynamically Typed (Python) |
|---|---|---|
| Declare types? | Yes, required | No, automatic |
| Can a variable change type? | No | Yes |
| Catches type mistakes | Before running | Only when that line runs |
| Code length | More verbose | Shorter, faster to write |
| Best for | Large, safety-critical systems | Quick development, flexibility |

**The trade-off in one sentence:** static typing catches more mistakes early but makes you write more, while dynamic typing is faster and more flexible but can let type-related bugs slip through until runtime.

### Where Python Shines, and Where It Struggles

**Where Python shines**

- **Readability and beginner-friendliness.** Clean syntax means more time solving your problem and less time fighting the language.
- **Data science, machine learning and AI.** The ecosystem (NumPy, Pandas, scikit-learn, TensorFlow, PyTorch) is unmatched.
- **Automation and scripting.** Renaming a thousand files, scraping a website, automating a repetitive task.
- **Web development (back-end).** Django and Flask power many real sites and APIs.
- **Glue code and rapid prototyping.** Quickly stitching systems together and testing an idea before committing to a bigger build.
- **The community and libraries.** For almost any task, someone has already built a free library for it.

**Where Python struggles**

- **Raw speed.** Being interpreted and dynamically typed makes Python much slower than C++ or Rust for heavy number-crunching. (Speed-critical libraries like NumPy are written in C underneath, so you get C-like speed with Python-like ease.)
- **Mobile app development.** Rarely used for phone apps; Swift and Kotlin dominate.
- **Memory usage.** Python generally uses more memory than lower-level languages.
- **True parallel processing.** The Global Interpreter Lock (GIL) has long stopped standard Python running threads at full speed across multiple cores. Python 3.13 added an experimental free-threaded mode and 3.14 made it officially supported, but it is still opt-in and the ecosystem is catching up.
- **Very high-performance or real-time systems.** Operating systems, game engines and anything where every microsecond counts are better served by compiled languages.

### Putting It All Together

Let's trace what happens with two simple lines:

```python
age = 25
print(age)
```

1. You write these two human-readable lines.
2. Python **parses** them, checking the grammar is valid.
3. It translates them into **bytecode**.
4. In memory, an object holding `25` is created and the label `age` is attached to it.
5. The **PVM** executes the bytecode line by line, first the assignment, then the print.
6. The **CPU** carries out the actual work, and `25` appears on your screen.
7. When the program ends, the **garbage collector** cleans up the memory automatically.

All of that, from two tiny lines, happens in a fraction of a second.
